```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
iris = datasets.load_iris(as_frame = True).frame
iris.head()
```

```python
iris.info()
```

```python
iris.describe()
```

```python
iris.head(2)
```

```python
# Splitting the preprocessed iris dataset into X and y
X = iris.drop("target",axis=1)
y = iris["target"]
X.head()
```

```python
# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
X_train.head()
```

```python
# For Support Vector Classifier - Scale the input data
# For Support Vector Regressor - Scale both input and output data 

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svc = SVC()
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report, f1_score
print("Accuracy : ",accuracy_score(y_test, y_pred))
print("classification_report : ",classification_report(y_test, y_pred))
```

```python
# Linear Kernel
svc = SVC(kernel="linear")
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)
print("Accuracy : ",accuracy_score(y_test, y_pred))
```

```python
# Polynomial Kernel
svc = SVC(kernel="poly")
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)
print("Accuracy : ",accuracy_score(y_test, y_pred))
```

```python
# Sigmoid Kernel
svc = SVC(kernel="sigmoid")
svc.fit(X_train_scaled, y_train)

y_pred = svc.predict(X_test_scaled)
print("Accuracy : ",accuracy_score(y_test, y_pred))
```

```python
# Hyper Parameter Turning
C_vals = [0.5, 1, 2, 3, 4, 5, 6]

for c_val in C_vals:
    svc = SVC(kernel="rbf", C=c_val)
    svc.fit(X_train_scaled, y_train)
    
    y_pred = svc.predict(X_test_scaled)
    print("C = ",c_val,"& Accuracy : ",accuracy_score(y_test, y_pred))
    
```
